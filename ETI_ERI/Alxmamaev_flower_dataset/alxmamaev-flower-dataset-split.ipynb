```python
import numpy as np
import os
import tensorflow as tf
import cv2
import json
import matplotlib.pyplot as plt
import pandas as pd
import random

print("Hello world.")
```

```python

daisy_folder = "/kaggle/input/datasets/alxmamaev/flowers-recognition/flowers/daisy"
images_length  = len(os.listdir(daisy_folder))
print(images_length)


images = os.listdir(daisy_folder)
random.shuffle(images)
daisy_split = []

for i in range(0, images_length):
    if i < 611:
        daisy_split.append([images[i] , "training"])
    elif i < 687:
        daisy_split.append([images[i], "validation"])
    else:
        daisy_split.append([images[i], "testing"])

daisy_split = pd.DataFrame(daisy_split, columns=['image', 'used_for'])
daisy_split.to_csv('daisy_split.csv')


print("Hello world.")

# 506 64 
```
