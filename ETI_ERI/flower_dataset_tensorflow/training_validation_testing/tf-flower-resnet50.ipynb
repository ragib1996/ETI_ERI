```python
x = np.asarray(x)
x_validation = np.asarray(x_validation)
x_test = np.asarray(x_test)

print(len(x)," ", len(x_validation)," ", len(x_test))
```

```python
from tensorflow.keras.utils import to_categorical 

label = np.array(label)
label = to_categorical(label , 5 )
print(len(label))



label_validation = np.array(label_validation)
label_validation = to_categorical(label_validation , 5 )
print(len(label_validation))

label_test = np.array(label_test)
label_test = to_categorical(label_test , 5 )
print(len(label_test))
```

```python
from tensorflow.keras.applications import ResNet50
from tensorflow.keras.layers import Input, GlobalAveragePooling2D, Dense, Dropout
from tensorflow.keras.models import Model
from tensorflow.keras.optimizers import Adam

# Define input layer
inputs = Input(shape=(224, 224, 3))

# Load ResNet50 base model without top layers
base_model = ResNet50(weights="imagenet", include_top=False, input_tensor=inputs)
base_model.trainable = False  # Freeze base model

# Add custom layers on top
y = base_model.output
y = GlobalAveragePooling2D()(y)
y = Dropout(0.3)(y)
y = Dense(128, activation="relu")(y)
outputs = Dense( 5, activation="softmax")(y)

# Create the functional model
model = Model(inputs=inputs, outputs=outputs)

# Compile the model
model.compile(
    optimizer=Adam(learning_rate=5e-5),
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()
```

```python

checkpoint1 = tf.keras.callbacks.ModelCheckpoint(
    "ResNet50_tf_flowers.keras",
    monitor="val_accuracy",
    save_best_only=True,
    mode="max",
    verbose=1
)

history1 = model.fit(
    x, label,
    validation_data=(x_validation, label_validation),
    epochs=100,
    batch_size=16,
    callbacks=[checkpoint1]
)
```

```python

test_loss, test_acc = model.evaluate(x_test, label_test)
print("Test Accuracy:", test_acc)
```
