```python
import numpy as np
import os
import tensorflow as tf
import cv2
import json
import matplotlib.pyplot as plt
import pandas as pd
import random

print("Hello world.")
```

```python
apple_folder = "/kaggle/input/datasets/mahmoudshaheen1134/plant-leaf-image-dataset/New_data/Apple"
images_length  = len(os.listdir(apple_folder))
print(images_length)


images = os.listdir(apple_folder)
random.shuffle(images)
apple_split = []

for i in range(0, images_length):
    if i < 415:
        apple_split.append([images[i] , "training"])
    elif i < 467:
        apple_split.append([images[i], "validation"])
    else:
        apple_split.append([images[i], "testing"])

apple_split = pd.DataFrame(apple_split, columns=['image', 'used_for'])
apple_split.to_csv('apple_split.csv')
```
